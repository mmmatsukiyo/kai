```python
# 4 --- Prime Star Systems ---
```

```python
arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])
```

```python
def is_prime(n):
    if n < 2:
        return False
    for j in range(2, int(n**0.5) + 1):
        if n % j == 0:
            return False
    return True

def row_has_prime(row):
    for n in row:
        if is_prime(n):
            return True
    return False

def primefinder(arr):
    rows = []
    for row in arr:
        if row_has_prime(row):
            rows.append(row)
    return np.array(rows)

print(primefinder(arr))
```

```python
# 5 --- Let's Play Checkers! ---
```

```python
def checkerboard():
    # 5.1
    board = np.zeros((8, 8))

    # 5.2
    board[0::2, 0::2] = 1

    # 5.3
    board[1::2, 1::2] = 1

    return board

print(checkerboard())
```

```python
# 5.4
def reverse_checkerboard():
    board = np.zeros((8, 8))
    board[0::2, 1::2] = 1   # rows 0, 2, 4, 6: 0 1 0 1 0 1 0 1
    board[1::2, 0::2] = 1   # rows 1, 3, 5, 7: 1 0 1 0 1 0 1 0
    return board

print(reverse_checkerboard())
```

```python
# 6 --- The Expanding Universe ---
```

```python
def expansion(arr, space):
    spaces = ""
    i = 0
    while i < space:
        spaces += " "
        i += 1
    
    new_list = []
    for i in arr:
        new_string = ""
        for letter in i:
            new_string += letter + spaces
        new_list.append(new_string)

        new_arr = np.array(new_list)
    return new_arr

universe = np.array(["galaxy", "clusters"])

print(expansion(universe, 2))
```

```python
# 7 --- Second-Brightest Star ---
```

```python
stars = np.random.randint(500, 2000, (5, 5))
print(stars)

def second_brightest(arr):
    result = []
    for row in arr:
        sorted_row = sorted(row)
        result.append(sorted_row[-2])
    return np.array(result)

print(second_brightest(stars))
```
