```python
# 2.6
print(df["latitude"].median())
print(df["longitude"].median())
```

```python
# 2.7
lat_abs = abs(df["latitude"] - df["latitude"].median())
print(lat_abs)
df["dist to lat med"] = lat_abs
```

```python
# 2.8
long_abs = abs(df["longitude"] - df["longitude"].median())
print(long_abs)
df["dist to long med"] = long_abs
```

```python
# 2.9
print(df)
```

```python
print(df.shape)
```

```python
# 52 rows, 6 columns
```

```python
# 2.10
closest_med_lat = df["dist to lat med"].idxmin()
print(f"Closest to median latitude: {df["name"][closest_med_lat]}, {df["dist to lat med"][closest_med_lat]:.2f}")

farthest_med_lat = df["dist to lat med"].idxmax()
print(f"Farthest to median latitude: {df["name"][farthest_med_lat]}, {df["dist to lat med"][farthest_med_lat]:.2f}")

closest_med_long = df["dist to long med"].idxmin()
print(f"Closest to median longitude: {df["name"][closest_med_long]}, {df["dist to long med"][closest_med_long]:.2f}")

farthest_med_long = df["dist to long med"].idxmax()
print(f"Farthest to median longitude: {df["name"][farthest_med_long]}, {df["dist to long med"][farthest_med_long]:.2f}")
```

```python
# 3.1
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 3.4
pen_df = sns.load_dataset("penguins")
```

```python
# 3.5
pen_df = pen_df.dropna()
```

```python
# 3.6
pen_df.shape
```

```python
# 333 rows, 7 columns
```

```python
pen_df.head(5)
```

```python
pen_df.tail(5)
```

```python
print(pen_df.columns.tolist())
```

```python
# species: penguin species
# island: island penguin inhabits
# bill_length_mm: penguin bill length in mm
# bill_depth_mm: penguin bill depth in mm
# flipper_length_mm: penguiin flipper length in mm
# body_mass_g: penguin body mass in g
# sex: penguin sex (male or female)
```

```python
# 3.7

# subplots
fig, axes = plt.subplots(2, 2,figsize=(10,10))

# scatter 1
axes[0, 0].scatter(pen_df["bill_length_mm"], pen_df["bill_depth_mm"])

# scatter 2
axes[0, 1].scatter(pen_df["flipper_length_mm"], pen_df["body_mass_g"])

# histogram
sns.histplot(data=pen_df, x="body_mass_g", ax=axes[1, 0])

# categorical
sns.boxplot(data=pen_df, x="species", y="flipper_length_mm", ax=axes[1, 1])

plt.show()
```

```python
# 3.8
fig, axes = plt.subplots(2, 2,figsize=(10,10))

species_list = pen_df["species"].unique()

for species in species_list:
    df_subset = pen_df[pen_df["species"] == species]
    axes[0, 0].scatter(df_subset["bill_length_mm"], df_subset["bill_depth_mm"], label=f"{species}")

# scatter 1
axes[0, 0].set_title("Bill Length vs. Bill Depth")
axes[0, 0].set_xlabel("Bill Length (mm)")
axes[0, 0].set_ylabel("Bill Depth (mm)")
axes[0, 0].legend()

# scatter 2
for species in species_list:
    df_subset = pen_df[pen_df["species"] == species]
    axes[0, 1].scatter(df_subset["flipper_length_mm"], df_subset["body_mass_g"], label=f"{species}")
axes[0, 1].set_title("Flipper Length vs. Body Mass")
axes[0, 1].set_xlabel("Flipper Length (mm)")
axes[0, 1].set_ylabel("Body Mass (g)")
axes[0, 1].legend()

# histogram
sns.histplot(data=pen_df, x="body_mass_g", ax=axes[1, 0])
axes[1, 0].set_title("Distribution of Body Mass")
axes[1, 0].set_xlabel("Body Mass (g)")
axes[1, 0].set_ylabel("Count")

# categorical
sns.boxplot(data=pen_df, x="species", y="flipper_length_mm", ax=axes[1, 1])
axes[1, 1].set_title("Flipper Length by Species")
axes[1, 1].set_xlabel("Species")
axes[1, 1].set_ylabel("Flipper Length (mm)")

plt.show()
```

```python
# 3.9

# scatter 1:
# - The data shows the bill length vs bill depth of each species of penguin, showing the size difference in beaks by species

# scatter 2:
# - The data shows the flipper length vs body mass and the length needed depending on the penguin species' weight

# histogram:
# - The data shows the distribution of the body mass of all the penguins regardless of species in grams

# categorical plot:
# - The data shows the mean, median, min, and max of each species' flipper length

# I learned that the dataset provides plenty of information to work with when comparing different aspects of penguins
```
